# Project 03-01: Mad Libs Story Generator

A Mad Libs game: the user supplies words (names, adjectives, numbers, ...) and the
program drops them into a story template.

> **Run order:** helper functions are defined first, then the story templates,
> then `STORIES`, then `main()`. Use **Kernel -> Restart & Run All** to run it cleanly.

## Helper Functions

`print_style()` draws a separator line; `get_word()` asks for a word and keeps
asking until it is valid (letters only, at least `min_length` characters).

```python
def print_style(char="#", length=60):
    print(char * length)
```

```python
def get_word(prompt, min_length=2):
    while True:
        word = input(prompt).strip()
        if len(word) >= min_length and word.isalpha():
            return word.title()
        else:
            print(f"  Please enter a real word (letters only, at least {min_length} chars).\n")
```

## Story 1 - Ghost

```python
def story_ghost(words):
    return f"""
About {words['years']} years ago, a woman named {words['fname']} lived in a small village called {words['village']}.
She was killed by the villagers for studying {words['subject']}. After her death, villagers started being murdered.
Many people believed that the soul of the poor woman was responsible for those killings.
How {words['adjective']}!
"""
```

## Story 2 - King

```python
def story_king(words):
    return f"""
Many years ago, in the kingdom of {words['kingdom']}, there lived a king named {words['king_name']}. 
He was famous for his {words['adjective']} rule and his love of {words['activity']}. 
One day, a mysterious {words['animal']} appeared near the royal palace, causing fear among the villagers. 
Determined to uncover the truth, the king gathered {words['number']} brave knights and set out on a journey. 
After traveling through the {words['place']}, they discovered that the strange event was caused by a giant {words['animal']}. 
With great courage, King {words['king_name']} used a magical {words['object']} to save the kingdom. How {words['adjective2']}!
"""
```

## Variable Declaration (the `STORIES` menu)

```python
# STORIES = { "1" : (a, b, [ (c), (d) ] ), "2" : (m, n, [ (o), (p) ] ) } : a dictionary
# (a, b, [ (c), (d) ] ) and (m, n, [ (o), (p) ] ): tuples
# [ (c), (d) ] and [ (o), (p) ] : lists
# (c), (d), (o) and (p) : tuples
STORIES = {
    "1" : ("ghost", story_ghost, [
            ("years", "number of years (digit only)"),
            ("fname", "a female name"),
            ("village", "a village name"),
            ("subject", "a studying subject"),
            ("adjective", "a negative adjective")
            ]
            ),

    "2" : ("king", story_king, [
            ("kingdom", "name of a kingdom"),
            ("king_name", "the name for king"),
            ("adjective", "a positive adjective"),
            ("activity", "an activity"),
            ("animal", "a ferocious animal name"),
            ("number", "a number (digit only)"),
            ("place", "a place name"),
            ("object", "an object name"),
            ("adjective2", "another positive adjective")
            ]
            )
}
```

## Main Function

```python
def main():

    # optional design part
    print_style()
    print("\t\t  welcome to my story")
    print_style()

    # lists the story
    print("Available stories:")
    for key, (story_title, _, _) in STORIES.items():
        print(f"   {key}. {story_title.title()}")
    print()

    # choose the story
    choice = input("Choose a story number: ").strip()
    # making sure to choose the correct story number
    while choice not in STORIES.keys():                   # STORIES and STORIES.keys() : same!
        print("Invalid choice. Please try again.")
        choice = input("Choose a story number: ").strip()

    # slicing dictionaries
    # choice -> keys
    # (story_title, story_func, word_prompts) -> values
    # story_title -> str type, story_func -> a variable, word_prompts -> list of tuples
    (story_title, story_func, word_prompts) = STORIES[choice]

    # story title
    print()
    print_style("-")
    print(f"  Story: {story_title.title()}")
    print("  Fill in each blank")
    print_style("-")
    print()

    # collect words from user
    collected_words = {}  # empty dictionary

    for (key, description) in word_prompts:
        if key in ["years", "number"]:
            while True:
                val = input(f" Enter {description}: ").strip()
                if val.isdigit():
                    collected_words[key] = val
                    break
                print("  Please enter digits only.\n")
        else:
            collected_words[key] = get_word(f" Enter {description}: ")

    # Build and display the story
    print()
    print_style("-")
    print(f"\t\t YOUR MAD LIBS STORY")
    print_style("-")

    finished_story = story_func(collected_words)
    print(finished_story)
    print_style("#")

    # Offer to play again
    again = input("Play again? (yes / no): ").strip().lower()
    if again in ("yes", "y"):
        print()
        main()   # recursive call to restart
    else:
        print("\nThanks for playing! Goodbye.")
```

```python
if __name__ == "__main__":
    main()
```
