# Project 03-02: BMI Calculator & Health Advisor

Calculates BMI from a user-supplied weight and height (with selectable units),
classifies it into a WHO health category, and prints advice plus the healthy
weight range for that height.

> **Run order:** cells go top-to-bottom (constants -> helpers -> conversions ->
> calculation -> display -> menu -> `main`). Use **Kernel -> Restart & Run All**
> to run it cleanly.

## Constants

```python
# BMI boundary values
BMI_UNDERWEIGHT  = 18.5
BMI_NORMAL_HIGH  = 25.0
BMI_OVERWEIGHT   = 30.0

# Conversion factors
KG_PER_LB   = 0.453592    # 1 pound = 0.453592 kg
M_PER_INCH  = 0.0254      # 1 inch  = 0.0254 metres
CM_PER_M    = 100.0
```

## Helper Functions

`print_style()` draws a separator line; `get_positive_float()` keeps asking until
the user enters a number greater than zero.

```python
def print_style(char="=", width=65):
    print(char * width)
```

```python
def get_positive_float(prompt):
    while True:
        try:
            value = float(input(prompt))
            if value > 0:
                return value
            else:
                print("  Value must be greater than zero. Try again.\n")
        except ValueError:
            print("  Please enter a valid number. Try again.\n")
```

## Unit Conversion

Convert whichever unit the user picks into the base units the BMI formula needs
(kilograms and metres).

```python
def mass_to_kg(mass, selected_mass):
    if selected_mass == "kg":
        return mass
    elif selected_mass == "lb":
        return mass * KG_PER_LB
```

```python
def height_to_m(height, selected_height):
    if selected_height == "meter":
        return height
    elif selected_height == "centimeter":
        return height / CM_PER_M
    elif selected_height == "inch":
        return height * M_PER_INCH
```

## BMI Calculation

```python
def calculate_bmi(mass_kg, height_m):
    bmi = mass_kg / (height_m ** 2)
    return round(bmi, 1)
```

```python
def bmi_category(bmi):
    if bmi < BMI_UNDERWEIGHT:
        category = "Underweight"
        advice   = ("Consider eating more nutrient-dense meals. "
                    "Consult a nutritionist for a personalised plan.")
    elif bmi < BMI_NORMAL_HIGH:
        category = "Normal weight"
        advice   = ("Great work! Maintain your current lifestyle — "
                    "balanced diet and regular exercise.")
    elif bmi < BMI_OVERWEIGHT:
        category = "Overweight"
        advice   = ("Some lifestyle adjustments may help. "
                    "Focus on whole foods and regular physical activity.")
    else:
        category = "Obese"
        advice   = ("It is advisable to consult a healthcare professional "
                    "for a structured weight-management programme.")

    return category, advice
```

```python
def healthy_weight_range(height_m):
    min_mass = round(BMI_UNDERWEIGHT * (height_m ** 2), 1)
    max_mass = round(BMI_NORMAL_HIGH * (height_m ** 2), 1)
    return min_mass, max_mass
```

## Display Results

```python
def display_results(name, mass_kg, height_m, bmi, category, advice):
    min_w, max_w = healthy_weight_range(height_m)

    print()
    print_style()
    print(f"  BMI Report for {name}")
    print_style()
    print(f"  Weight   : {mass_kg:.1f} kg  ({mass_kg / KG_PER_LB:.1f} lbs)")
    print(f"  Height   : {height_m:.2f} m  ({height_m * CM_PER_M:.0f} cm)")
    print(f"  BMI      : {bmi}")
    print(f"  Category : {category}")
    print_style("-")
    print(f"  Advice   : {advice}")
    print_style("-")
    print(f"  Healthy weight range for your height:")
    print(f"    {min_w} kg – {max_w} kg")
    print_style("-")
    print()
```

## Unit Options (the menu)

```python
UNIT = {
    "mass" : ("kg", "lb"),
    "height" : ("meter", "centimeter", "inch")
}
```

## Main Function

```python
def main():

    # Headline
    print_style()
    print("\t\tBMI Calculator & Health Advisor")
    print_style()
    print()

    # User info
    name = input("Your name: ").strip().title() or "User"

    """
    name = input("Your name: ")
        if name:
            name = name.strip().title()
        else:
            name = 'User'
    """

    # Choose unit system and value
    # Mass unit
    print("\nMass Unit System:")
    total_mass_units = len(UNIT["mass"])

    for serial, mass in enumerate(UNIT["mass"], start=1):
        print(f"{serial}. {mass}")

    while True:
        choice = input("Choose preferred mass serial number: ").strip()
        if choice.isdigit():
            mass_choice = int(choice)
            if 1 <= mass_choice <= total_mass_units:
                break
        print(f"Invalid choice. Please enter a number between 1 and {total_mass_units}.")

    selected_mass = UNIT["mass"][mass_choice - 1]
    mass = get_positive_float(f"Enter mass in {selected_mass}: ")
    mass_kg = mass_to_kg(mass, selected_mass)

    # Height unit
    print("\nHeight Unit System:")
    total_height_units = len(UNIT["height"])

    for serial, height in enumerate(UNIT["height"], start=1):
        print(f"{serial}. {height}")

    while True:
        choice = input("Choose preferred height serial number: ").strip()
        if choice.isdigit():
            height_choice = int(choice)
            if 1 <= height_choice <= total_height_units:
                break
        print(f"Invalid choice. Please enter a number between 1 and {total_height_units}.")

    selected_height = UNIT["height"][height_choice - 1]
    height = get_positive_float(f"Enter height in {selected_height}: ")
    height_m = height_to_m(height, selected_height)

    # Calculate
    bmi = calculate_bmi(mass_kg, height_m)
    category, advice = bmi_category(bmi)

    # Display results
    display_results(name, mass_kg, height_m, bmi, category, advice)

    # Option to repeat
    again = input("Calculate another? (yes / no): ").strip().lower()
    if again in ("yes", "y"):
        print()
        main()
    else:
        print("Stay healthy!")
```

```python
if __name__ == "__main__":
    main()
```
